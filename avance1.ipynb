## 1. Pregunta de investigación

### Contexto

La duración de la estadía hospitalaria es un indicador central del uso de camas y del costo de la atención. En Chile, FONASA financia la mayor parte de las hospitalizaciones de la red pública, y una parte se paga por Grupos Relacionados por el Diagnóstico (GRD), que asignan un valor según la complejidad esperada del caso. Las estadías muy largas concentran una fracción importante de los días-cama. Conocer qué características de los pacientes y de su diagnóstico se asocian a estas estadías ayuda a planificar camas y a evaluar los mecanismos de pago.

### Pregunta

> **¿Se asocian el diagnóstico principal (capítulo CIE-10), la intervención quirúrgica, la edad y el sexo con la duración de la estadía hospitalaria y con la ocurrencia de estadías prolongadas (más de 15 días) en los egresos de beneficiarios FONASA en Chile durante 2020?**

- **Estadía prolongada:** más de 15 días de estada, que corresponde al percentil 90 de la distribución de días de estada de los egresos FONASA 2020.
- La pregunta es medible: todas las variables están en la base de egresos del DEIS y cada asociación se contrasta con una prueba estadística (sección 5).

### Hipótesis

- **H₀:** La duración de la estadía y la ocurrencia de estadías prolongadas son independientes del capítulo diagnóstico CIE-10, de la intervención quirúrgica, de la edad y del sexo del paciente.
- **H₁:** Al menos uno de estos factores (capítulo diagnóstico CIE-10, intervención quirúrgica, edad o sexo) se asocia a la duración de la estadía o a la ocurrencia de estadías prolongadas.

## 2. Revisión de literatura

### Síntesis crítica

Los Grupos Relacionados por el Diagnóstico (GRD) nacieron de agrupar pacientes con diagnósticos y procedimientos parecidos que consumen recursos parecidos, y la estadía fue la variable que se usó para formar los grupos (Fetter et al., 1980). La idea de fondo es que el diagnóstico principal y la cirugía explican buena parte de la duración. Freitas et al. (2012) muestran que eso no alcanza: con nueve millones de episodios de hospitales públicos portugueses entre 2000 y 2009, y con umbrales de estadía extrema definidos dentro de cada GRD, la edad, el tipo de ingreso y el tipo de hospital seguían asociados a las estadías extremas. Esos casos eran solo el 3,9 % de los episodios, pero sumaban el 19,2 % de los días de hospitalización. Parte de la duración tampoco tiene una causa clínica: Rojas-García et al. (2018) revisan el alta retrasada, que afecta sobre todo a adultos mayores que ya no necesitan estar hospitalizados, pero esperan cupo en cuidados posteriores.

Para el análisis estadístico importa la forma de la variable. Los días de estada son conteos positivos con una cola larga a la derecha: Marazzi et al. (1998) los ajustan con distribuciones asimétricas (lognormal, Weibull y gamma) y Kulinskaya et al. (2005) proponen métodos robustos para compararlos entre hospitales. Con una distribución así, la media y las pruebas que suponen normalidad pierden confiabilidad, y por eso aquí se verifica la normalidad antes de elegir cada prueba (sección 5.1).

En cuanto a la predicción, la revisión de Lequertier et al. (2021) encontró que la regresión fue el método más usado para predecir la estadía (64,9 % de los casos), aunque el *machine learning* ganó terreno en la última década. Stone et al. (2022) revisan 93 estudios y llegan a un diagnóstico parecido: los datos, las variables y las métricas cambian tanto de un trabajo a otro que los resultados son difíciles de comparar, y proponen un marco común. Esto respalda comparar un modelo interpretable (regresión logística) con uno flexible (Random Forest) sobre los mismos datos y con las mismas métricas.

Por último, con bases administrativas de cientos de miles de registros casi cualquier diferencia resulta significativa. Lin et al. (2013) recomiendan leer el valor-p junto con el tamaño del efecto, así que en la sección 5 cada prueba se acompaña de una medida de magnitud, interpretada con los umbrales de Cohen (1988).

### Gap identificado

En Chile, los trabajos sobre estadía hospitalaria con datos públicos se concentran en poblaciones específicas. Olivares-Tirado (2020), con egresos DEIS de 2017, estudió la hospitalización social de personas de 60 años o más: el 1 % de sus egresos correspondió a estadías de 60 días o más con baja necesidad de atención médica, y esos egresos concentraron el 15 % de los días-cama. En la búsqueda realizada no se encontraron análisis que evalúen en conjunto el diagnóstico, la cirugía, la edad y el sexo como factores de estadía prolongada en el total de egresos FONASA. Además, 2020 fue el primer año de la pandemia de COVID-19 y no está descrito cómo se repartieron ese año las estadías prolongadas entre diagnósticos. Este proyecto aborda ese vacío con los 968.098 egresos FONASA de 2020 publicados por el DEIS.

## 3. Metodología

### Tipo de análisis

- **Descriptivo:** caracterizar la distribución de los días de estada y la proporción de estadías prolongadas por grupo.
- **Inferencial:** contrastar las hipótesis de asociación entre las variables (sección 5).
- **Predictivo:** clasificar los egresos en estadía prolongada o no prolongada a partir de las variables seleccionadas.

### Fuente de datos, unidad de análisis y período

| Aspecto | Detalle |
|---|---|
| Fuente | Departamento de Estadísticas e Información de Salud (DEIS), MINSAL. Datos abiertos de egresos hospitalarios: `EGRESOS_2020.zip` (https://repositoriodeis.minsal.cl/DatosAbiertos/EGRESOS/) |
| Registros | 1.330.477 egresos en 2020; **968.495 de beneficiarios FONASA** (`PREVISION = 1`), dentro del rango exigido de 700.000 a 1.000.000. Se excluyen 397 egresos con más de 365 días de estada por ser valores implausibles (máximo registrado: 16.598 días); quedan 968.098 |

### Variables seleccionadas

| Variable | Columna original | Tipo | Justificación |
|---|---|---|---|
| Días de estada (`dias_estada`) | `DIAS_ESTADA` | Numérica | Variable respuesta: duración de la hospitalización |
| Estadía prolongada (`prolongada`) | Derivada: `DIAS_ESTADA` > 15 | Categórica (Sí / No) | Variable respuesta binaria; identifica el 10 % de estadías más largas |
| Edad (`edad`) | `GRUPO_EDAD` (punto medio del tramo) | Numérica | La edad aumenta la estadía según la literatura (Freitas et al., 2012). La base solo entrega tramos de 10 años, por lo que se usa el punto medio |
| Sexo (`sexo`) | `SEXO` | Categórica (Hombre / Mujer) | Variable demográfica básica |
| Intervención quirúrgica (`cirugia`) | `INTERV_Q` (1 = sí, 2 = no) | Categórica (Sí / No) | Distingue egresos quirúrgicos de médicos; los GRD usan el procedimiento para agrupar |
| Capítulo CIE-10 (`capitulo`) | `DIAG1`, agrupado con el diccionario del DEIS | Categórica (21 capítulos) | Aproxima la complejidad clínica del caso, ya que la base abierta no incluye el peso GRD |

### Modelos propuestos

- **Regresión logística** para la estadía prolongada (Sí / No). Es un modelo interpretable: entrega razones de odds para cada factor.
- **Random Forest** para la estadía prolongada. Puede captar relaciones no lineales (por ejemplo, la forma de la relación con la edad) e interacciones entre variables.

### Métricas de evaluación

- **AUC-ROC:** capacidad del modelo para distinguir estadías prolongadas de no prolongadas.
- **Sensibilidad (recall), precisión y F1:** relevantes porque la clase "prolongada" es minoritaria (cerca del 10 %).
- **Matriz de confusión** sobre un conjunto de prueba (partición 70 % entrenamiento / 30 % prueba, estratificada).

## 5. Análisis de relaciones entre variables

Para cada prueba se reportan las hipótesis (H₀ / H₁), el estadístico, el valor-p, el nivel de significancia (**α = 0,05**), la decisión y la interpretación.

**Nota sobre los valores-p:** con n ≈ 968.000, varios valores-p son tan pequeños que el computador los muestra como 0. Se reportan como "p < 0,001".

**Carga de los datos y construcción de las variables.**

````python
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.stats.diagnostic import lilliefors

warnings.filterwarnings("ignore")
ALPHA = 0.05
rng = np.random.default_rng(2026)
DATA = Path("data/raw/x2020")

df = pd.read_csv(DATA / "EGRE_DATOS_ABIERTOS_2020.csv", sep=";", encoding="cp1252", dtype=str)
print("Egresos 2020 (total):", f"{len(df):,}".replace(",", "."))
df = df[df["PREVISION"] == "1"].copy()  # beneficiarios FONASA
print("Egresos FONASA:      ", f"{len(df):,}".replace(",", "."))
df["dias_estada"] = pd.to_numeric(df["DIAS_ESTADA"])
df = df[df["dias_estada"] <= 365].copy()  # se excluyen estadías > 1 año (valores implausibles)
print("Con estada <= 365 d: ", f"{len(df):,}".replace(",", "."))

# Variables seleccionadas
EDAD_MEDIA = {"menor de un año": 0.5, "1 a 9": 5, "10 a 19": 14.5, "20 a 29": 24.5, "30 a 39": 34.5,
              "40 a 49": 44.5, "50 a 59": 54.5, "60 a 69": 64.5, "70 a 79": 74.5, "80 a 89": 84.5,
              "90 y más": 92.5}
df["edad"] = df["GRUPO_EDAD"].map(EDAD_MEDIA)
df["sexo"] = df["SEXO"].str.capitalize()
df["cirugia"] = df["INTERV_Q"].map({"1": "Sí", "2": "No"})
cie = pd.read_excel(DATA / "Diccionario BD egresos hospitalario.xlsx", "codigo CIE-10", header=8)
cie.columns = cie.columns.str.strip()
df["capitulo"] = (df["DIAG1"].map(dict(zip(cie["CODIGO SUBCATEGORIA"], cie["CAPITULO"])))
                  .fillna(df["DIAG1"].str[:3].map(dict(zip(cie["CODIGO CATEGORIA"], cie["CAPITULO"])))))
df["prolongada"] = np.where(df["dias_estada"] > 15, "Sí", "No")

def decision(p):
    return "Se rechaza H0" if p < ALPHA else "No se rechaza H0"

df[["dias_estada", "edad", "sexo", "cirugia", "capitulo", "prolongada"]].head()
````

### 5.1 Pruebas de bondad de ajuste (supuesto de normalidad)

Variables numéricas: `dias_estada` y `edad`.

- **H₀:** la variable sigue una distribución normal.
- **H₁:** la variable no sigue una distribución normal.
- α = 0,05.

Shapiro–Wilk se aplica a una muestra aleatoria de 5.000 egresos, porque la prueba no es válida para muestras mucho mayores. Kolmogorov–Smirnov con corrección de Lilliefors y Anderson–Darling se aplican a los 968.098 egresos.

````python
def normalidad(x, nombre):
    x = np.asarray(x, float)
    sw = stats.shapiro(rng.choice(x, 5000, replace=False))  # Shapiro-Wilk: muestra aleatoria n = 5.000
    d, p_ks = lilliefors(x, dist="norm", pvalmethod="approx")
    ad = stats.anderson(x, dist="norm")
    return {"Variable": nombre,
            "Shapiro-Wilk W": sw.statistic, "p (SW)": sw.pvalue,
            "Lilliefors D": d, "p (KS)": p_ks,
            "Anderson-Darling A²": ad.statistic, "Valor crítico AD (5%)": ad.critical_values[2],
            "Decisión": decision(min(sw.pvalue, p_ks)) if ad.statistic > ad.critical_values[2] else "Revisar"}

tabla_normalidad = pd.DataFrame([normalidad(df["dias_estada"], "dias_estada"),
                                 normalidad(df["edad"], "edad")])

fig, ax = plt.subplots(2, 2, figsize=(12, 8))
ax[0, 0].hist(df["dias_estada"].clip(upper=60), bins=60, color="#4C72B0")
ax[0, 0].set_title("Histograma: días de estada (recortado en 60)")
ax[0, 1].hist(df["edad"], bins=11, color="#4C72B0")
ax[0, 1].set_title("Histograma: edad (punto medio del tramo)")
stats.probplot(rng.choice(df["dias_estada"].values, 5000, replace=False), plot=ax[1, 0])
ax[1, 0].set_title("Q-Q normal: días de estada")
stats.probplot(rng.choice(df["edad"].values, 5000, replace=False), plot=ax[1, 1])
ax[1, 1].set_title("Q-Q normal: edad")
fig.tight_layout(); plt.show()
tabla_normalidad
````

**Resultados**

| Variable | Shapiro–Wilk | K–S (Lilliefors) | Anderson–Darling | Decisión |
|---|---|---|---|---|
| `dias_estada` | W = 0,397; p < 0,001 | D = 0,328; p < 0,001 | A² = 150.921 > 0,752 (valor crítico 5 %) | Se rechaza H₀ |
| `edad` | W = 0,956; p < 0,001 | D = 0,129; p < 0,001 | A² = 13.546 > 0,752 | Se rechaza H₀ |

**Interpretación.** Las tres pruebas rechazan la normalidad en ambas variables. Los días de estada tienen una fuerte asimetría a la derecha: la mayoría de las hospitalizaciones dura entre 1 y 7 días (mediana 3), pero hay una cola larga que llega hasta 365 días. En el gráfico Q-Q, los puntos se alejan de la recta en el extremo derecho. La edad tampoco es normal: tiene dos concentraciones (adultos de 20 a 39 años, por los partos, y adultos mayores) y solo está disponible en tramos.

**Consecuencia:** como no se cumple el supuesto de normalidad, en las secciones 5.2 y 5.3 se usan las pruebas no paramétricas: **Spearman** en lugar de Pearson, **Mann–Whitney U** en lugar de t de Student y **Kruskal–Wallis** en lugar de ANOVA.

### 5.2 Numérica vs. numérica: edad vs. días de estada

Como ninguna de las dos variables es normal (sección 5.1), no se cumple el supuesto de Pearson y se usa la **correlación de Spearman (ρ)**.

- **H₀:** ρ = 0 (no hay relación monótona entre la edad y los días de estada).
- **H₁:** ρ ≠ 0.
- α = 0,05.

````python
rho, p = stats.spearmanr(df["edad"], df["dias_estada"])
print(f"Spearman: rho = {rho:.4f} | p = {p:.3g} | n = {len(df)}")
print(decision(p))

mediana_por_edad = df.groupby("edad")["dias_estada"].median()
muestra = df.sample(20000, random_state=2026)
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(muestra["edad"] + rng.uniform(-3, 3, len(muestra)), muestra["dias_estada"], s=4, alpha=0.25,
           label="egresos (muestra de 20.000)")
ax.plot(mediana_por_edad.index, mediana_por_edad.values, "r-o", label="mediana por tramo")
ax.set_yscale("log"); ax.set_xlabel("Edad (años, punto medio del tramo)"); ax.set_ylabel("Días de estada (escala log)")
ax.set_title("Edad vs. días de estada"); ax.legend(); plt.show()
mediana_por_edad
````

**Resultados**

| Coeficiente | Valor | Valor-p | n | Decisión |
|---|---|---|---|---|
| Spearman ρ | 0,250 | < 0,001 | 968.098 | Se rechaza H₀ |

**Interpretación.** Existe una relación **positiva y débil** entre la edad y los días de estada: a mayor edad, la hospitalización tiende a ser más larga. La mediana de estada es de 2 días entre 1 y 39 años y sube a 5 días entre 60 y 79 años y a 6 días desde los 80 años. La relación no es del todo monótona: los menores de un año (recién nacidos) tienen una mediana de 4 días, más alta que la de niños y adultos jóvenes. Que la correlación sea débil indica que la edad por sí sola explica poco de la duración, por lo que debe analizarse junto con el diagnóstico.

### 5.3 Numérica vs. categórica: días de estada según sexo, cirugía y capítulo CIE-10

Como los días de estada no son normales (sección 5.1), se usan las alternativas no paramétricas: **Mann–Whitney U** para dos grupos (sexo, cirugía) y **Kruskal–Wallis** para tres o más grupos (capítulo CIE-10).

**Hipótesis (α = 0,05):**

| Variable | H₀ | H₁ |
|---|---|---|
| Sexo | La distribución de los días de estada es igual en hombres y mujeres | La distribución difiere entre hombres y mujeres |
| Cirugía | La distribución de los días de estada es igual con y sin intervención quirúrgica | La distribución difiere |
| Capítulo CIE-10 | La distribución de los días de estada es igual en los 21 capítulos | Al menos un capítulo tiene una distribución distinta |

````python
def mann_whitney(var, a, b):
    xa = df.loc[df[var] == a, "dias_estada"]; xb = df.loc[df[var] == b, "dias_estada"]
    u = stats.mannwhitneyu(xa, xb, alternative="two-sided")
    return {"Variable": var, "Grupos": f"{a} / {b}", "n": f"{len(xa)} / {len(xb)}",
            "Mediana": f"{xa.median():.0f} / {xb.median():.0f}", "Prueba": "Mann-Whitney U",
            "Estadístico": u.statistic, "p": u.pvalue, "Decisión": decision(u.pvalue)}

grupos_cap = [g["dias_estada"].values for _, g in df.groupby("capitulo")]
kw = stats.kruskal(*grupos_cap)
tabla_num_cat = pd.DataFrame([
    mann_whitney("sexo", "Hombre", "Mujer"),
    mann_whitney("cirugia", "Sí", "No"),
    {"Variable": "capitulo", "Grupos": f"{len(grupos_cap)} capítulos", "n": len(df), "Mediana": "ver tabla",
     "Prueba": "Kruskal-Wallis H", "Estadístico": kw.statistic, "p": kw.pvalue, "Decisión": decision(kw.pvalue)},
])

fig, ax = plt.subplots(1, 3, figsize=(18, 6), gridspec_kw={"width_ratios": [1, 1, 3]})
for a_, var, orden in [(ax[0], "sexo", ["Hombre", "Mujer"]), (ax[1], "cirugia", ["Sí", "No"])]:
    a_.boxplot([df.loc[df[var] == o, "dias_estada"] for o in orden], tick_labels=orden, showfliers=False)
    a_.set_title(f"Días de estada por {var}"); a_.set_ylabel("Días")
orden_cap = df.groupby("capitulo")["dias_estada"].median().sort_values().index
ax[2].boxplot([df.loc[df["capitulo"] == c, "dias_estada"] for c in orden_cap], tick_labels=orden_cap, showfliers=False)
ax[2].tick_params(axis="x", rotation=60); ax[2].set_title("Días de estada por capítulo CIE-10")
fig.suptitle("Boxplots (sin valores atípicos)"); fig.tight_layout(); plt.show()

display(df.groupby("capitulo")["dias_estada"].agg(n="size", mediana="median", media="mean")
        .sort_values("mediana", ascending=False).round(1))
tabla_num_cat
````

**Resultados**

| Variable | Grupos (n) | Mediana (días) | Prueba | Estadístico | Valor-p | Decisión |
|---|---|---|---|---|---|---|
| Sexo | Hombre (387.368) / Mujer (580.730) | 4 / 3 | Mann–Whitney U | U = 1,28 × 10¹¹ | < 0,001 | Se rechaza H₀ |
| Cirugía | Sí (362.148) / No (605.950) | 2 / 4 | Mann–Whitney U | U = 8,51 × 10¹⁰ | < 0,001 | Se rechaza H₀ |
| Capítulo CIE-10 | 21 capítulos (968.098) | 1 a 9 | Kruskal–Wallis | H = 116.411 (gl = 20) | < 0,001 | Se rechaza H₀ |

**Interpretación.**
- **Sexo:** los hombres tienen estadías algo más largas que las mujeres (mediana 4 vs. 3 días). Una explicación probable es que una gran parte de las hospitalizaciones de mujeres son partos (capítulo O), que son cortos (mediana 2 días).
- **Cirugía:** los egresos con intervención quirúrgica son más cortos (mediana 2 vs. 4 días). Muchas cirugías son programadas y de recuperación rápida, mientras que las hospitalizaciones médicas (infecciones, enfermedades cardiovasculares y respiratorias, COVID-19) suelen requerir más días de tratamiento.
- **Capítulo CIE-10:** es la variable con mayores diferencias. Los trastornos mentales (F00-F99) y el COVID-19 (U00-U99) tienen una mediana de 9 días, y las enfermedades infecciosas, circulatorias, respiratorias, endocrinas y de la piel, una mediana de 5 días. En el otro extremo, los partos, la patología digestiva y las enfermedades musculoesqueléticas tienen una mediana de 2 días, y las enfermedades del ojo y los controles de salud (Z), de 1 día. Esto apoya usar el diagnóstico como principal aproximación de la complejidad clínica.

### 5.4 Categórica vs. categórica: estadía prolongada según sexo, cirugía y capítulo CIE-10

Se usa la prueba **chi-cuadrado de independencia** sobre las tablas de contingencia de cada variable con `prolongada` (Sí / No).

- **H₀:** la variable y la estadía prolongada son independientes.
- **H₁:** la variable y la estadía prolongada están asociadas.
- α = 0,05.

**Supuesto:** todas las frecuencias esperadas deben ser ≥ 5. Si en una tabla 2 × 2 hubiera frecuencias esperadas pequeñas, se usaría el test exacto de Fisher.

````python
def chi_cuadrado(var):
    tabla = pd.crosstab(df[var], df["prolongada"])
    chi2, p, gl, esperadas = stats.chi2_contingency(tabla)
    v = np.sqrt(chi2 / (tabla.values.sum() * (min(tabla.shape) - 1)))
    display(tabla.assign(**{"% prolongada": (100 * tabla["Sí"] / tabla.sum(axis=1)).round(1)}))
    return {"Variable": var, "Chi²": chi2, "gl": gl, "p": p, "V de Cramér": v,
            "Mín. frec. esperada": esperadas.min(), "Decisión": decision(p)}

tabla_chi2 = pd.DataFrame([chi_cuadrado(v) for v in ["sexo", "cirugia", "capitulo"]])

fig, ax = plt.subplots(1, 3, figsize=(18, 6), gridspec_kw={"width_ratios": [1, 1, 3]})
for a_, var in zip(ax, ["sexo", "cirugia", "capitulo"]):
    pct = pd.crosstab(df[var], df["prolongada"], normalize="index") * 100
    if var == "capitulo":
        pct = pct.sort_values("Sí")
    pct.plot(kind="bar", stacked=True, ax=a_, color=["#9ecae1", "#de2d26"], legend=(var == "sexo"))
    a_.set_title(f"Estadía prolongada por {var}"); a_.set_ylabel("%"); a_.set_xlabel("")
ax[2].tick_params(axis="x", rotation=60)
fig.tight_layout(); plt.show()
tabla_chi2
````

**Verificación del supuesto.** La menor frecuencia esperada es 36.376 (sexo), 34.007 (cirugía) y 175 (capítulo CIE-10). Todas son mayores que 5, por lo que el chi-cuadrado es válido y no se requiere el test exacto de Fisher.

**Resultados**

| Variable | % prolongadas por grupo | χ² | gl | Valor-p | V de Cramér | Decisión |
|---|---|---|---|---|---|---|
| Sexo | Hombre 13,1 % / Mujer 6,9 % | 10.332,2 | 1 | < 0,001 | 0,103 | Se rechaza H₀ |
| Cirugía | Sí 7,4 % / No 10,6 % | 2.749,3 | 1 | < 0,001 | 0,053 | Se rechaza H₀ |

**Interpretación.**
- **Capítulo CIE-10:** es la asociación más fuerte (V = 0,25, débil a moderada). La proporción de estadías prolongadas varía mucho según el diagnóstico: 35,6 % en trastornos mentales, 26,2 % en COVID-19, 19,1 % en enfermedades infecciosas y 17,2 % en afecciones perinatales, frente a menos de 3 % en partos, enfermedades del ojo y del oído, síntomas (R) y controles (Z). Hay grupos diagnósticos claramente asociados a estadías prolongadas.
- **Sexo:** los hombres tienen casi el doble de estadías prolongadas que las mujeres (13,1 % vs. 6,9 %), pero la asociación es débil (V = 0,10) y, como en 5.3, puede deberse en parte al gran número de partos entre las mujeres.
- **Cirugía:** la asociación es significativa pero muy débil (V = 0,05). Los egresos sin cirugía tienen algo más de estadías prolongadas (10,6 % vs. 7,4 %). Con casi un millón de registros, una diferencia de esta magnitud resulta significativa, pero tiene poca relevancia práctica.
