## 1. Pregunta de investigación

### Contexto

La duración de la estadía hospitalaria es uno de los indicadores más usados para medir el uso de camas y el costo de la atención. En Chile, FONASA financia la mayor parte de las hospitalizaciones de la red pública, y parte de ese pago se hace por Grupos Relacionados por el Diagnóstico (GRD), que asignan un valor según la complejidad esperada del caso. En los egresos FONASA de 2020, el 9,4 % que duró más de 15 días sumó el 47,1 % de los días-cama (sección 5). Saber qué características del paciente y de su diagnóstico se asocian a esas estadías sirve para planificar camas y para revisar si el pago por GRD refleja el costo real de los casos largos.

### Pregunta

> **¿Se asocian el diagnóstico principal (capítulo CIE-10), la intervención quirúrgica, la edad y el sexo con la duración de la estadía hospitalaria y con la ocurrencia de estadías prolongadas (más de 15 días) en los egresos de beneficiarios FONASA en Chile durante 2020?**

Se considera prolongada la estadía de más de 15 días, que es el percentil 90 de los días de estada de los egresos FONASA 2020 y abarca al 9,4 % de ellos. Todas las variables están en la base de egresos del DEIS, y cada asociación se contrasta con una prueba estadística en la sección 5.

### Hipótesis

- **H₀:** la duración de la estadía y la ocurrencia de estadías prolongadas son independientes del capítulo diagnóstico CIE-10, de la intervención quirúrgica, de la edad y del sexo.
- **H₁:** al menos uno de estos factores se asocia a la duración de la estadía o a la ocurrencia de estadías prolongadas.

La hipótesis general se separa en una hipótesis específica por factor:

| Hipótesis | H₁ específica | Dirección esperada |
|---|---|---|
| H1a | El capítulo CIE-10 se asocia a la duración y a la estadía prolongada | Sin dirección fija |
| H1b | La intervención quirúrgica se asocia a la duración y a la estadía prolongada | Sin dirección fija |
| H1c | La edad se asocia a la duración y a la estadía prolongada | Más días a mayor edad (Freitas et al., 2012) |
| H1d | El sexo se asocia a la duración y a la estadía prolongada | Sin dirección fija |

Con cerca de un millón de egresos, rechazar H₀ es casi seguro aunque la diferencia sea mínima. Por eso cada hipótesis se juzga también por el tamaño del efecto, y un factor se considera relevante si su efecto es al menos pequeño según los umbrales de Cohen (1988).

## 2. Revisión de literatura

### Síntesis crítica

Los Grupos Relacionados por el Diagnóstico (GRD) nacieron de agrupar pacientes con diagnósticos y procedimientos parecidos que consumen recursos parecidos, y la estadía fue la variable que se usó para formar los grupos (Fetter et al., 1980). La idea de fondo es que el diagnóstico principal y la cirugía explican buena parte de la duración. Freitas et al. (2012) muestran que eso no alcanza: con nueve millones de episodios de hospitales públicos portugueses entre 2000 y 2009, y con umbrales de estadía extrema definidos dentro de cada GRD, la edad, el tipo de ingreso y el tipo de hospital seguían asociados a las estadías extremas. Esos casos eran solo el 3,9 % de los episodios, pero sumaban el 19,2 % de los días de hospitalización. Parte de la duración tampoco tiene una causa clínica: Rojas-García et al. (2018) revisan el alta retrasada, que afecta sobre todo a adultos mayores que ya no necesitan estar hospitalizados, pero esperan cupo en cuidados posteriores.

Los días de estada son conteos positivos con una cola larga a la derecha: Marazzi et al. (1998) los ajustan con distribuciones asimétricas (lognormal, Weibull y gamma) y Kulinskaya et al. (2005) proponen métodos robustos para compararlos entre hospitales. Con esa forma, la media y las pruebas que suponen normalidad pierden confiabilidad, y por eso aquí se verifica la normalidad antes de elegir cada prueba (sección 5.1).

En cuanto a la predicción, la revisión de Lequertier et al. (2021) encontró que la regresión fue el método más usado para predecir la estadía (64,9 % de los casos), aunque el *machine learning* ganó terreno en la última década. Stone et al. (2022) revisan 93 estudios y llegan a un diagnóstico parecido: los datos, las variables y las métricas cambian tanto de un trabajo a otro que los resultados son difíciles de comparar, y proponen un marco común. Esto respalda comparar un modelo interpretable (regresión logística) con uno flexible (Random Forest) sobre los mismos datos y con las mismas métricas.

Por último, con bases administrativas de cientos de miles de registros casi cualquier diferencia resulta significativa. Lin et al. (2013) recomiendan leer el valor-p junto con el tamaño del efecto, así que en la sección 5 cada prueba se acompaña de una medida de magnitud, interpretada con los umbrales de Cohen (1988).

### Gap identificado

En Chile, los trabajos sobre estadía hospitalaria con datos públicos se concentran en poblaciones específicas. Olivares-Tirado (2020), con egresos DEIS de 2017, estudió la hospitalización social de personas de 60 años o más: el 1 % de sus egresos correspondió a estadías de 60 días o más con baja necesidad de atención médica, y esos egresos concentraron el 15 % de los días-cama. En la búsqueda realizada no se encontraron análisis que evalúen en conjunto el diagnóstico, la cirugía, la edad y el sexo como factores de estadía prolongada en el total de egresos FONASA. Además, 2020 fue el primer año de la pandemia de COVID-19 y no está descrito cómo se repartieron ese año las estadías prolongadas entre diagnósticos. Este proyecto aborda ese vacío con los 968.098 egresos FONASA de 2020 publicados por el DEIS.

## 3. Metodología

### Tipo de análisis

El estudio es descriptivo, inferencial y predictivo. Es descriptivo porque caracteriza la distribución de los días de estada y la proporción de estadías prolongadas en cada grupo; inferencial porque contrasta las hipótesis de la sección 1 con las pruebas de la sección 5, y predictivo porque en el Avance 2 se ajustarán modelos para clasificar los egresos en prolongados o no y para estimar los días de estada.

### Fuente de datos, unidad de análisis y período

| Aspecto | Detalle |
|---|---|
| Fuente | Departamento de Estadísticas e Información de Salud (DEIS), MINSAL. Datos abiertos de egresos hospitalarios: `EGRESOS_2020.zip` (https://repositoriodeis.minsal.cl/DatosAbiertos/EGRESOS/) |
| Registros | 1.330.477 egresos en 2020, de los cuales 968.495 son de beneficiarios FONASA (`PREVISION = 1`). Se excluyen 397 egresos con más de 365 días de estada, que se consideran errores de registro (el máximo es de 16.598 días, unos 45 años). Quedan 968.098 egresos, dentro del rango exigido de 700.000 a 1.000.000 |
| Unidad de análisis | El egreso hospitalario |
| Período | 1 de enero al 31 de diciembre de 2020 |

### Variables seleccionadas

| Variable | Columna original | Tipo | Justificación |
|---|---|---|---|
| Días de estada (`dias_estada`) | `DIAS_ESTADA` | Numérica (conteo) | Variable respuesta: duración de la hospitalización |
| Estadía prolongada (`prolongada`) | Derivada: `DIAS_ESTADA` > 15 | Categórica (Sí / No) | Variable respuesta binaria; marca al 9,4 % de egresos con las estadías más largas (sobre el percentil 90) |
| Edad (`edad`, `tramo_edad`) | `GRUPO_EDAD` | Ordinal (11 tramos) | La literatura asocia más edad con estadías más largas (Freitas et al., 2012). La base solo entrega tramos de 10 años: para la correlación se usa el punto medio de cada tramo y para la tabla de contingencia, el tramo |
| Sexo (`sexo`) | `SEXO` | Categórica (Hombre / Mujer) | Variable demográfica básica |
| Intervención quirúrgica (`cirugia`) | `INTERV_Q` (1 = sí, 2 = no) | Categórica (Sí / No) | Separa egresos quirúrgicos de médicos; los GRD también usan el procedimiento para agrupar |
| Capítulo CIE-10 (`capitulo`) | `DIAG1`, agrupado con el diccionario del DEIS | Categórica (21 capítulos) | Aproxima la complejidad clínica del caso, porque la base abierta no incluye el peso GRD |

### Modelos propuestos

| Modelo | Variable respuesta | Por qué |
|---|---|---|
| Regresión logística | Estadía prolongada | Entrega razones de odds ajustadas. Permite ver, por ejemplo, si la diferencia entre hombres y mujeres se mantiene al controlar por diagnóstico (los partos son cortos y solo afectan a mujeres) |
| Random Forest | Estadía prolongada | Capta relaciones no lineales, como la de la edad, que no es monótona, e interacciones entre variables |
| Regresión binomial negativa | Días de estada | Los días son un conteo con varianza mucho mayor que la media (media 6,7 y desviación estándar 12,8), por lo que un modelo de Poisson no es adecuado |

### Métricas de evaluación

Los modelos de clasificación se evalúan con AUC-ROC, AUC-PR, sensibilidad, precisión, F1 y la matriz de confusión. La AUC-PR se incluye porque la clase "prolongada" es minoritaria (9,4 %), y en ese caso la AUC-ROC puede verse alta aunque el modelo detecte pocos casos. El modelo de días de estada se evalúa con MAE y RMSE. En todos los casos se usa una partición estratificada de 70 % para entrenamiento y 30 % para prueba, y los hiperparámetros del Random Forest se ajustan con validación cruzada de 5 particiones dentro del conjunto de entrenamiento.

### Limitaciones del diseño

La base abierta del DEIS no trae el peso relativo GRD, el hospital, la fecha de ingreso ni los diagnósticos secundarios. Por eso la complejidad clínica se aproxima con el capítulo CIE-10 y la cirugía, y no se puede responder directamente si el peso GRD explica la duración. La edad solo viene en tramos de 10 años. Además, 2020 fue un año atípico por la pandemia, lo que limita la generalización; las bases 2019 y 2021 a 2024 permitirán comparar en el Avance 2.

## 5. Análisis de relaciones entre variables

Para cada prueba se informan las hipótesis (H₀ / H₁), el estadístico, el valor-p, el nivel de significancia (α = 0,05), la decisión, un tamaño de efecto y la interpretación en el contexto del problema.

Con n ≈ 968.000, varios valores-p son tan pequeños que el computador los muestra como 0; en el texto se informan como "p < 0,001". Ese mismo tamaño de muestra hace que casi cualquier diferencia resulte significativa, así que la conclusión de cada prueba se apoya en el tamaño del efecto, leído con los umbrales de Cohen (1988): para la V de Cramér y las correlaciones (ρ o r), 0,1 es un efecto pequeño, 0,3 mediano y 0,5 grande; para ε², 0,01 es pequeño, 0,06 mediano y 0,14 grande.

En total se hacen 8 pruebas de asociación sobre los mismos datos. Todas tienen p < 0,001, por lo que una corrección por comparaciones múltiples (Bonferroni, con α = 0,05 / 8 ≈ 0,006) no cambia ninguna decisión. La corrección sí se aplica en el post-hoc de 5.3, donde se comparan 210 pares de capítulos.

#### Carga de los datos y construcción de las variables

```python
import warnings
import zipfile
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.stats.diagnostic import lilliefors
from statsmodels.stats.multitest import multipletests

warnings.filterwarnings("ignore")
ALPHA = 0.05
rng = np.random.default_rng(2026)
DATA = Path("data/raw")
FUENTE = "Fuente: elaboración propia con datos de egresos hospitalarios DEIS-MINSAL, 2020."

def miles(n):
    return f"{n:,}".replace(",", ".")

def fuente(fig):
    fig.text(0.01, 0, FUENTE, fontsize=9, style="italic", va="top")

with zipfile.ZipFile(DATA / "EGRESOS_2020.zip") as z:
    df = pd.read_csv(z.open("EGRE_DATOS_ABIERTOS_2020.csv"), sep=";", encoding="cp1252", dtype=str)
print("Egresos 2020 (total):", miles(len(df)))
df = df[df["PREVISION"] == "1"].copy()  # beneficiarios FONASA
print("Egresos FONASA:      ", miles(len(df)))
df["dias_estada"] = pd.to_numeric(df["DIAS_ESTADA"])
print("Estadías > 365 d:    ", miles((df["dias_estada"] > 365).sum()), "| máximo:", miles(df["dias_estada"].max()), "días")
df = df[df["dias_estada"] <= 365].copy()  # se excluyen estadías > 1 año (errores de registro)
print("Con estada <= 365 d: ", miles(len(df)))

# Variables seleccionadas
EDAD_MEDIA = {"menor de un año": 0.5, "1 a 9": 5, "10 a 19": 14.5, "20 a 29": 24.5, "30 a 39": 34.5,
              "40 a 49": 44.5, "50 a 59": 54.5, "60 a 69": 64.5, "70 a 79": 74.5, "80 a 89": 84.5,
              "90 y más": 92.5}
df["edad"] = df["GRUPO_EDAD"].map(EDAD_MEDIA)
df["tramo_edad"] = pd.Categorical(df["GRUPO_EDAD"], categories=list(EDAD_MEDIA), ordered=True)
df["sexo"] = df["SEXO"].str.capitalize()
df["cirugia"] = df["INTERV_Q"].map({"1": "Sí", "2": "No"})
cie = pd.read_excel(DATA / "x2020/Diccionario BD egresos hospitalario.xlsx", "codigo CIE-10", header=8)
cie.columns = cie.columns.str.strip()
df["capitulo"] = (df["DIAG1"].map(dict(zip(cie["CODIGO SUBCATEGORIA"], cie["CAPITULO"])))
                  .fillna(df["DIAG1"].str[:3].map(dict(zip(cie["CODIGO CATEGORIA"], cie["CAPITULO"])))))
df["prolongada"] = np.where(df["dias_estada"] > 15, "Sí", "No")

larga = df["prolongada"] == "Sí"
print(f"\nDías de estada: media {df['dias_estada'].mean():.1f} | desv. estándar {df['dias_estada'].std():.1f} | "
      f"mediana {df['dias_estada'].median():.0f} | percentil 90 {df['dias_estada'].quantile(0.9):.0f}")
print(f"Estadías prolongadas (> 15 días): {larga.mean():.1%} de los egresos y "
      f"{df.loc[larga, 'dias_estada'].sum() / df['dias_estada'].sum():.1%} de los días-cama")

def decision(p):
    return "Se rechaza H0" if p < ALPHA else "No se rechaza H0"

df[["dias_estada", "edad", "tramo_edad", "sexo", "cirugia", "capitulo", "prolongada"]].head()
```

### 5.1 Pruebas de bondad de ajuste (supuesto de normalidad)

Variables numéricas: `dias_estada` y `edad`.

- **H₀:** la variable sigue una distribución normal.
- **H₁:** la variable no sigue una distribución normal.
- α = 0,05.

Shapiro-Wilk se aplica a una muestra aleatoria de 5.000 egresos, porque con muestras más grandes el valor-p que entrega scipy deja de ser exacto. Kolmogorov-Smirnov con corrección de Lilliefors y Anderson-Darling se aplican a los 968.098 egresos. Se rechaza la normalidad cuando las tres pruebas la rechazan.

```python
def normalidad(x, nombre):
    x = np.asarray(x, float)
    sw = stats.shapiro(rng.choice(x, 5000, replace=False))  # Shapiro-Wilk: muestra aleatoria n = 5.000
    d, p_ks = lilliefors(x, dist="norm", pvalmethod="approx")
    ad = stats.anderson(x, dist="norm")
    rechaza = sw.pvalue < ALPHA and p_ks < ALPHA and ad.statistic > ad.critical_values[2]
    return {"Variable": nombre,
            "Shapiro-Wilk W": sw.statistic, "p (SW)": sw.pvalue,
            "Lilliefors D": d, "p (KS)": p_ks,
            "Anderson-Darling A²": ad.statistic, "Valor crítico AD (5%)": ad.critical_values[2],
            "Decisión": "Se rechaza H0" if rechaza else "No se rechaza H0"}

tabla_normalidad = pd.DataFrame([normalidad(df["dias_estada"], "dias_estada"),
                                 normalidad(df["edad"], "edad")])

fig, ax = plt.subplots(2, 2, figsize=(12, 8))
ax[0, 0].hist(df["dias_estada"].clip(upper=60), bins=60, color="#4C72B0")
ax[0, 0].set_title("Histograma: días de estada (recortado en 60)")
ax[0, 0].set_xlabel("Días de estada"); ax[0, 0].set_ylabel("Egresos")
df["tramo_edad"].value_counts(sort=False).plot(kind="bar", ax=ax[0, 1], color="#4C72B0", width=0.9)
ax[0, 1].set_title("Egresos por tramo de edad")
ax[0, 1].set_xlabel("Tramo de edad (años)"); ax[0, 1].set_ylabel("Egresos")
plt.setp(ax[0, 1].get_xticklabels(), rotation=45, ha="right")
stats.probplot(rng.choice(df["dias_estada"].values, 5000, replace=False), plot=ax[1, 0])
ax[1, 0].set_title("Q-Q normal: días de estada")
stats.probplot(rng.choice(df["edad"].values, 5000, replace=False), plot=ax[1, 1])
ax[1, 1].set_title("Q-Q normal: edad")
for a_ in ax[1]:
    a_.set_xlabel("Cuantiles teóricos (normal)"); a_.set_ylabel("Valores observados")
fig.suptitle("Figura 2. Histogramas y gráficos Q-Q de los días de estada y la edad")
fig.tight_layout(); fuente(fig); plt.show()
tabla_normalidad
```

| Variable | Shapiro-Wilk | K-S (Lilliefors) | Anderson-Darling | Decisión |
|---|---|---|---|---|
| `dias_estada` | W = 0,397; p < 0,001 | D = 0,328; p < 0,001 | A² = 150.921 > 0,752 (valor crítico 5 %) | Se rechaza H₀ |
| `edad` | W = 0,956; p < 0,001 | D = 0,129; p < 0,001 | A² = 13.546 > 0,752 | Se rechaza H₀ |

Las tres pruebas rechazan la normalidad en ambas variables. Los días de estada tienen una asimetría fuerte a la derecha: la mitad de las hospitalizaciones dura 3 días o menos y el 90 % dura 15 o menos, pero la cola llega a 365 días. En el gráfico Q-Q de la Figura 2 los puntos se despegan de la recta en el extremo superior. La edad tampoco es normal: tiene un máximo entre los 20 y 39 años, en buena parte por los partos, y otro entre los 60 y 69, y además solo existe en tramos, lo que se ve como escalones en su gráfico Q-Q.

Como no se cumple el supuesto, en 5.2 y 5.3 se usan pruebas no paramétricas: Spearman en lugar de Pearson, Mann-Whitney U en lugar de t de Student y Kruskal-Wallis en lugar de ANOVA.

### 5.2 Numérica vs. numérica: edad vs. días de estada

Como ninguna de las dos variables es normal (sección 5.1), no se cumple el supuesto de Pearson y se usa la correlación de Spearman (ρ). La edad viene en tramos, pero Spearman trabaja con rangos, así que usar el punto medio de cada tramo no cambia el resultado.

- **H₀:** ρ = 0 (no hay relación monótona entre la edad y los días de estada).
- **H₁:** ρ ≠ 0.
- α = 0,05.

```python
rho, p = stats.spearmanr(df["edad"], df["dias_estada"])
print(f"Spearman: rho = {rho:.4f} | p = {p:.3g} | n = {miles(len(df))}")
print(decision(p))

mediana_por_edad = df.groupby("edad")["dias_estada"].median()
muestra = df.sample(20000, random_state=2026)
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(muestra["edad"] + rng.uniform(-3, 3, len(muestra)), muestra["dias_estada"], s=4, alpha=0.25,
           label="egresos (muestra de 20.000)")
ax.plot(mediana_por_edad.index, mediana_por_edad.values, "r-o", label="mediana por tramo")
ax.set_yscale("log"); ax.set_xlabel("Edad (años, punto medio del tramo)"); ax.set_ylabel("Días de estada (escala log)")
ax.set_title("Figura 3. Edad y días de estada"); ax.legend(); fig.tight_layout(); fuente(fig); plt.show()
mediana_por_edad
```

| Coeficiente | Valor | Valor-p | n | Decisión |
|---|---|---|---|---|
| Spearman ρ | 0,250 | < 0,001 | 968.098 | Se rechaza H₀ |

La edad y los días de estada tienen una relación positiva pero pequeña (ρ = 0,25): a mayor edad, la hospitalización tiende a durar más. La mediana es de 2 días entre 1 y 39 años, sube a 5 días entre 60 y 79 y llega a 6 desde los 80 (Figura 3). La relación no es del todo monótona, porque los menores de un año tienen una mediana de 4 días, más que los niños y adultos jóvenes; Spearman no capta ese repunte y por eso subestima un poco la asociación. Con un ρ de 0,25, la edad por sí sola explica poco de la duración y conviene analizarla junto con el diagnóstico.

### 5.3 Numérica vs. categórica: días de estada según sexo, cirugía y capítulo CIE-10

Como los días de estada no son normales (sección 5.1), se usan las alternativas no paramétricas: Mann-Whitney U para dos grupos (sexo, cirugía) y Kruskal-Wallis para tres o más grupos (capítulo CIE-10). El tamaño de efecto es la correlación biserial de rangos *r* para Mann-Whitney y ε² para Kruskal-Wallis. Kruskal-Wallis solo indica que al menos un capítulo difiere, así que después se aplica la prueba de Dunn (1964) a los 210 pares de capítulos, con corrección de Holm (1979) por comparaciones múltiples.

Hipótesis (α = 0,05):

| Variable | H₀ | H₁ |
|---|---|---|
| Sexo | La distribución de los días de estada es igual en hombres y mujeres | La distribución difiere entre hombres y mujeres |
| Cirugía | La distribución de los días de estada es igual con y sin intervención quirúrgica | La distribución difiere |
| Capítulo CIE-10 | La distribución de los días de estada es igual en los 21 capítulos | Al menos un capítulo tiene una distribución distinta |

```python
NOMBRE = {"tramo_edad": "tramo de edad", "sexo": "sexo", "cirugia": "cirugía", "capitulo": "capítulo CIE-10"}

def mann_whitney(var, a, b):
    xa = df.loc[df[var] == a, "dias_estada"]; xb = df.loc[df[var] == b, "dias_estada"]
    u = stats.mannwhitneyu(xa, xb, alternative="two-sided")
    r = 2 * u.statistic / (len(xa) * len(xb)) - 1  # correlación biserial de rangos
    return {"Variable": var, "Grupos": f"{a} / {b}", "n": f"{miles(len(xa))} / {miles(len(xb))}",
            "Mediana": f"{xa.median():.0f} / {xb.median():.0f}", "Prueba": "Mann-Whitney U",
            "Estadístico": u.statistic, "p": u.pvalue, "Efecto": f"r = {r:.3f}", "Decisión": decision(u.pvalue)}

grupos_cap = [g["dias_estada"].values for _, g in df.groupby("capitulo")]
kw = stats.kruskal(*grupos_cap)
eps2 = kw.statistic / (len(df) - 1)
tabla_num_cat = pd.DataFrame([
    mann_whitney("sexo", "Hombre", "Mujer"),
    mann_whitney("cirugia", "Sí", "No"),
    {"Variable": "capitulo", "Grupos": f"{len(grupos_cap)} capítulos", "n": miles(len(df)), "Mediana": "ver tabla",
     "Prueba": "Kruskal-Wallis H", "Estadístico": kw.statistic, "p": kw.pvalue, "Efecto": f"ε² = {eps2:.3f}",
     "Decisión": decision(kw.pvalue)},
])

fig, ax = plt.subplots(1, 3, figsize=(18, 6), gridspec_kw={"width_ratios": [1, 1, 3]})
for a_, var, orden in [(ax[0], "sexo", ["Hombre", "Mujer"]), (ax[1], "cirugia", ["Sí", "No"])]:
    a_.boxplot([df.loc[df[var] == o, "dias_estada"] for o in orden], tick_labels=orden, showfliers=False)
    a_.set_title(f"Por {NOMBRE[var]}"); a_.set_ylabel("Días de estada")
orden_cap = df.groupby("capitulo")["dias_estada"].median().sort_values().index
ax[2].boxplot([df.loc[df["capitulo"] == c, "dias_estada"] for c in orden_cap], tick_labels=orden_cap, showfliers=False)
ax[2].tick_params(axis="x", rotation=60); ax[2].set_title("Por capítulo CIE-10"); ax[2].set_ylabel("Días de estada")
fig.suptitle("Figura 4. Días de estada por sexo, cirugía y capítulo CIE-10 (sin valores atípicos)")
fig.tight_layout(); fuente(fig); plt.show()

display(df.groupby("capitulo")["dias_estada"].agg(n="size", mediana="median", media="mean")
        .sort_values("mediana", ascending=False).round(1))
tabla_num_cat
```

```python
# Post-hoc de Dunn (1964) para los 210 pares de capítulos, con corrección de Holm
rangos = df["dias_estada"].rank()
N = len(df)
t = df["dias_estada"].value_counts().values.astype(float)
s2 = N * (N + 1) / 12 - (t**3 - t).sum() / (12 * (N - 1))  # varianza de los rangos corregida por empates
g = rangos.groupby(df["capitulo"]).agg(["mean", "size"])
pares = []
for i, a in enumerate(g.index):
    for b in g.index[i + 1:]:
        z = (g.loc[a, "mean"] - g.loc[b, "mean"]) / np.sqrt(s2 * (1 / g.loc[a, "size"] + 1 / g.loc[b, "size"]))
        pares.append((a, b, z, 2 * stats.norm.sf(abs(z))))
dunn = pd.DataFrame(pares, columns=["Capítulo A", "Capítulo B", "z", "p"])
dunn["p (Holm)"] = multipletests(dunn["p"], method="holm")[1]
print(f"Dunn: {(dunn['p (Holm)'] < ALPHA).sum()} de {len(dunn)} pares de capítulos difieren (p Holm < {ALPHA})")
print("Pares sin diferencia significativa:")
dunn[dunn["p (Holm)"] >= ALPHA].round(3)
```

| Variable | Grupos (n) | Mediana (días) | Prueba | Estadístico | Valor-p | Tamaño de efecto | Decisión |
|---|---|---|---|---|---|---|---|
| Sexo | Hombre (387.368) / Mujer (580.730) | 4 / 3 | Mann-Whitney U | U = 1,28 × 10¹¹ | < 0,001 | r = 0,14 (pequeño) | Se rechaza H₀ |
| Cirugía | Sí (362.148) / No (605.950) | 2 / 4 | Mann-Whitney U | U = 8,51 × 10¹⁰ | < 0,001 | r = −0,22 (pequeño) | Se rechaza H₀ |
| Capítulo CIE-10 | 21 capítulos (968.098) | 1 a 9 | Kruskal-Wallis | H = 116.411 (gl = 20) | < 0,001 | ε² = 0,12 (mediano) | Se rechaza H₀ |
| Capítulo CIE-10 (post-hoc) | 210 pares | | Dunn con corrección de Holm | | | | 201 de 210 pares difieren |

El capítulo CIE-10 es la variable que más separa la duración (ε² = 0,12, efecto mediano). Los trastornos mentales (F) y el COVID-19 (U) tienen una mediana de 9 días; las enfermedades infecciosas, circulatorias, respiratorias, endocrinas y de la piel, de 5 días; los partos, la patología digestiva y las enfermedades musculoesqueléticas, de 2 días, y las enfermedades del ojo y los controles de salud (Z), de 1 día (Figura 4). Según la prueba de Dunn, 201 de los 210 pares de capítulos difieren entre sí, y los 9 que no difieren son capítulos con medianas parecidas, como infecciosas y circulatorias (A-B e I) o endocrinas, respiratorias y de la piel (E, J y L). Con diferencias tan repartidas, el capítulo sirve como aproximación a la complejidad clínica.

Los egresos quirúrgicos duran menos que los médicos (mediana de 2 y 4 días; r = −0,22). Una explicación razonable es que muchas cirugías son programadas y de recuperación rápida, mientras que las hospitalizaciones médicas, como infecciones, enfermedades cardiovasculares y respiratorias o COVID-19, requieren más días de tratamiento. Los hombres tienen un día más de mediana que las mujeres (4 y 3 días; r = 0,14). Esa diferencia probablemente se explica en parte por los partos, que son cerca de un tercio de los egresos de mujeres (197.020 de 580.730) y duran poco; la regresión logística del Avance 2 permitirá comprobarlo al controlar por diagnóstico.

### 5.4 Categórica vs. categórica: estadía prolongada según tramo de edad, sexo, cirugía y capítulo CIE-10

Se usa la prueba chi-cuadrado de independencia sobre la tabla de contingencia de cada variable con `prolongada` (Sí / No). Para la edad se usa el tramo (11 categorías).

- **H₀:** la variable y la estadía prolongada son independientes.
- **H₁:** la variable y la estadía prolongada están asociadas.
- α = 0,05.

El chi-cuadrado exige que todas las frecuencias esperadas sean ≥ 5. Si en una tabla 2 × 2 hubiera frecuencias esperadas pequeñas, se usaría el test exacto de Fisher. La fuerza de la asociación se mide con la V de Cramér; como `prolongada` es binaria, los umbrales de Cohen son 0,1 (pequeño), 0,3 (mediano) y 0,5 (grande).

```python
def chi_cuadrado(var):
    tabla = pd.crosstab(df[var], df["prolongada"])
    chi2, p, gl, esperadas = stats.chi2_contingency(tabla)
    v = np.sqrt(chi2 / (tabla.values.sum() * (min(tabla.shape) - 1)))
    display(tabla.assign(**{"% prolongada": (100 * tabla["Sí"] / tabla.sum(axis=1)).round(1)}))
    return {"Variable": var, "Chi²": chi2, "gl": gl, "p": p, "V de Cramér": v,
            "Mín. frec. esperada": esperadas.min(), "Decisión": decision(p)}

VARS_CAT = ["tramo_edad", "sexo", "cirugia", "capitulo"]
tabla_chi2 = pd.DataFrame([chi_cuadrado(v) for v in VARS_CAT])

fig, ax = plt.subplots(1, 4, figsize=(20, 6), gridspec_kw={"width_ratios": [2.2, 0.8, 0.8, 3]})
for a_, var in zip(ax, VARS_CAT):
    pct = pd.crosstab(df[var], df["prolongada"], normalize="index") * 100
    if var == "capitulo":
        pct = pct.sort_values("Sí")
    pct.plot(kind="bar", stacked=True, ax=a_, color=["#9ecae1", "#de2d26"], legend=(var == "tramo_edad"))
    a_.set_title(f"Por {NOMBRE[var]}"); a_.set_ylabel("% de egresos"); a_.set_xlabel("")
ax[0].legend(title="Prolongada", loc="lower left")
for a_ in (ax[0], ax[3]):
    a_.tick_params(axis="x", rotation=60)
fig.suptitle("Figura 5. Porcentaje de estadías prolongadas (> 15 días) por tramo de edad, sexo, cirugía y capítulo CIE-10")
fig.tight_layout(); fuente(fig); plt.show()
tabla_chi2
```

La menor frecuencia esperada es 1.408 (tramo de edad), 36.376 (sexo), 34.007 (cirugía) y 175 (capítulo CIE-10). Todas superan 5, así que el chi-cuadrado es válido y no se necesita el test exacto de Fisher.

| Tramo de edad | 3,5 % (20 a 29 años) a 15,4 % (menores de un año y 70 a 79 años) | 26.265,1 | 10 | < 0,001 | 0,165 (pequeño) | Se rechaza H₀ |
| Sexo | Hombre 13,1 % / Mujer 6,9 % | 10.332,2 | 1 | < 0,001 | 0,103 (pequeño) | Se rechaza H₀ |

El capítulo CIE-10 vuelve a ser el factor más asociado (V = 0,25, en el límite superior de un efecto pequeño). La proporción de estadías prolongadas es de 35,6 % en trastornos mentales, 26,2 % en COVID-19, 19,1 % en enfermedades infecciosas y 17,2 % en afecciones perinatales, y baja a menos de 3 % en partos, enfermedades del ojo y del oído, síntomas (R) y controles (Z) (Figura 5).

La edad le sigue (V = 0,16), con una forma de U que la correlación de 5.2 no capta: los menores de un año tienen 15,4 % de estadías prolongadas, la proporción baja a 3,5 % entre los 20 y 29 años y vuelve a subir hasta 15,4 % entre los 70 y 79.

En los hombres, el 13,1 % de los egresos son estadías prolongadas, frente al 6,9 % en las mujeres. La asociación es pequeña (V = 0,10) y, como se vio en 5.3, puede deberse en parte a los partos. La cirugía tiene una asociación despreciable (V = 0,05): 10,6 % de estadías prolongadas sin cirugía y 7,4 % con cirugía. Con casi un millón de registros esa diferencia resulta significativa, pero tiene poca relevancia práctica.
